```python
import os, numpy as np, pandas as pd, rasterio
from collections import Counter

TRAIN = "train"
meta_af = pd.read_csv(f"{TRAIN}/af/meta_af.csv")
meta_bs = pd.read_csv(f"{TRAIN}/bs/meta_bs.csv")
print("AF train:", len(meta_af), "BS train:", len(meta_bs))
print("AF fire_event_id unique:", meta_af["fire_event_id"].nunique())
print("BS fire_event_id unique:", meta_bs["fire_event_id"].nunique())

# AF: доля пикселей горения
pos, tot = 0, 0
for f in os.listdir(f"{TRAIN}/af/masks"):
    m = rasterio.open(f"{TRAIN}/af/masks/{f}").read(1)
    pos += (m > 0).sum(); tot += m.size
print(f"AF positive pixel ratio: {pos/tot:.6f}")

# BS: распределение классов
cnt = Counter()
for f in os.listdir(f"{TRAIN}/bs/masks"):
    m = rasterio.open(f"{TRAIN}/bs/masks/{f}").read(1)
    for c in range(4):
        cnt[c] += int((m == c).sum())
print("BS class distribution:", cnt)
```
